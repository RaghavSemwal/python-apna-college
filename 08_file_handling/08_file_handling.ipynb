## 1. What is File I/O?
Python can be used to perform operations on external files stored in persistent storage (SSD/HDD) such as creating, reading, updating, and deleting.

### RAM vs Storage:
* **RAM (Random Access Memory):** Fast execution memory where program variables are created. It is **volatile** (data is lost when the computer/program shuts down).
* **Storage (Files):** Non-volatile memory used to **persist** data permanently (e.g., `.txt`, `.csv`, `.png`, `.py`).

### File Types:
1. **Text Files:** Data stored as human-readable character streams (`.txt`, `.docx`, `.log`, `.py`).
2. **Binary Files:** Data stored in binary/byte format (`.png`, `.jpg`, `.mp4`, `.mov`).

## 2. File Modes Summary

| Mode | Name | Description | Pointer Position | Truncate (Wipe Data)? |
| :--- | :--- | :--- | :--- | :--- |
| `'r'` | Read (Default) | Opens file for reading only (File must exist). | Beginning | No |
| `'w'` | Write | Opens for writing. Creates file if missing; wipes existing data. | Beginning | **Yes** |
| `'a'` | Append | Opens for writing. Adds data at the end of the file. | End | No |
| `'r+'`| Read + Write | Overwrites from start without truncating. | Beginning | No |
| `'w+'`| Write + Read | Opens for reading & writing; truncates file. | Beginning | **Yes** |
| `'a+'`| Append + Read| Opens for reading & appending to end. | End | No |
| `'rb'`/`'wb'`| Binary | Binary read / write modes. | - | - |

````python
# 1. Writing to a new file using basic open() and close()
f = open("demo.txt", "w")
f.write("I am learning Python.\nThis is file I/O in Python.\n")
f.close() # Always close files to free memory resources!

# 2. Reading the entire file
f = open("demo.txt", "r")
data = f.read()
print("--- Entire File Content ---")
print(data)
f.close()

# 3. Reading character-by-character vs Line-by-line
f = open("demo.txt", "r")
print("First 5 characters:", f.read(5))
f.close()

f = open("demo.txt", "r")
print("Line 1:", f.readline(), end="")
print("Line 2:", f.readline(), end="")
f.close()
````

## 3. The `with` Syntax (Best Practice)
Using the `with` statement creates a **context manager** that automatically closes the file once the indented block finishes executing, even if errors occur.

```python
with open("filename.txt", "mode") as f:
    # Perform operations

````python


### 💻 Code Cell 2

# Reading using with syntax (auto-closes file)
with open("demo.txt", "r") as f:
    data = f.read()
    print("Read with 'with' block:\n", data)

# Appending data using 'a' mode
with open("demo.txt", "a") as f:
    f.write("Then I will move to Django / Machine Learning.\n")

with open("demo.txt", "r") as f:
    print("Content after appending:\n", f.read())
````

## 4. Deleting Files with `os.remove()`
Python does not have an in-built `f.delete()` method. We import the standard library module `os` and use `os.remove(filename)`.

````python
import os

# Create a temporary file to test deletion
with open("temp_sample.txt", "w") as f:
    f.write("This file is created only to test deletion.")

print("Does file exist before removal?", os.path.exists("temp_sample.txt"))

# Delete file using os.remove
if os.path.exists("temp_sample.txt"):
    os.remove("temp_sample.txt")
    print("temp_sample.txt deleted successfully!")
````

### Problem 1 & 2: Create a File & Replace Words
1. Create `practice.txt` with sample text.
2. Replace all occurrences of `"Java"` with `"Python"`.

````python
# Step 1: Create practice.txt
with open("practice.txt", "w") as f:
    f.write("Hi everyone\nwe are learning File I/O\nusing Java.\nI like programming in Java.\n")

print("--- Initial Content ---")
with open("practice.txt", "r") as f:
    print(f.read())

# Step 2: Function to replace Java with Python
def replace_word():
    with open("practice.txt", "r") as f:
        data = f.read()

    new_data = data.replace("Java", "Python")

    with open("practice.txt", "w") as f:
        f.write(new_data)
    print("Successfully replaced 'Java' with 'Python'!")

replace_word()

print("\n--- Updated Content ---")
with open("practice.txt", "r") as f:
    print(f.read())
````

### Problem 3 & 4: Word Search & Line Finder
* Search if the word `"learning"` exists in the file.
* Find the exact line number where the word occurs first (return `-1` if not found).

````python
# Check if word exists in file
def check_for_word(target_word):
    with open("practice.txt", "r") as f:
        data = f.read()

    if target_word in data:
        print(f"Found: '{target_word}' exists in the file.")
    else:
        print(f"Not Found: '{target_word}' does not exist.")

check_for_word("learning")
check_for_word("Golang")


# Find line number where word occurs first
def check_for_line(target_word):
    line_num = 1
    with open("practice.txt", "r") as f:
        while True:
            line = f.readline()
            if not line: # Reached end of file (empty string)
                break
            if target_word in line:
                print(f"Word '{target_word}' found on Line: {line_num}")
                return line_num
            line_num += 1

    print(f"Word '{target_word}' not found.")
    return -1

check_for_line("learning")
check_for_line("Python")
check_for_line("JavaScript")
````

### Problem 5: Parse Comma-Separated Numbers & Count Even Numbers
Given a file containing comma-separated numbers (e.g., `1, 2, 76, 84, 90, 101`), count how many are **even numbers**.

````python
# Create a numbers file
with open("numbers.txt", "w") as f:
    f.write("1, 2, 45, 55, 86, 76, 84, 90, 101, 104")

# Read, split, typecast, and count even numbers
with open("numbers.txt", "r") as f:
    data = f.read()

# Using split(",") to convert string into list of strings
num_strings = data.split(",")
print("Extracted strings:", num_strings)

even_count = 0
for val in num_strings:
    if int(val.strip()) % 2 == 0:
        even_count += 1

print("\nTotal Even Numbers in File:", even_count)

# Alternate Pythonic One-Liner (List Comprehension)
alt_even_count = sum(1 for val in data.split(",") if int(val.strip()) % 2 == 0)
print("Total Even Numbers (One-Liner):", alt_even_count)
````
