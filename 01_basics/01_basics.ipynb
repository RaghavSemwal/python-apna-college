### 📝 Cell 4 (Markdown): Variables & Identifiers


## 2. Variables & Rules for Naming Identifiers
* **Variable:** A named memory location used to store data values.
* **Assignment Operator (`=`):** Assigns the value on the right-hand side to the variable name on the left.

### Rules for Identifiers (Variable Names):
1. Can contain letters (`a-z`, `A-Z`), digits (`0-9`), and underscores (`_`).
2. **Cannot start with a digit** (e.g., `1variable` is invalid; `variable1` is valid).
3. **Cannot use special symbols** like `@`, `#`, `$`, `%`.
4. **Case-Sensitive:** `Name` and `name` are treated as two distinct variables.
5. **Cannot use Python Keywords** as variable names (e.g., `class`, `for`, `if`, `True`).

### 💻 Cell 5 (Code): Variable Declaration & `type()` Function

````python
# Declaring variables
name = "Shradha"
age = 23
price = 25.99
is_student = True
a = None

# Printing variables
print("Name:", name)
print("Age:", age)
print("Price:", price)

# Checking Data Types using type()
print("Type of name:", type(name))         # <class 'str'>
print("Type of age:", type(age))           # <class 'int'>
print("Type of price:", type(price))       # <class 'float'>
print("Type of is_student:", type(is_student)) # <class 'bool'>
print("Type of a:", type(a))               # <class 'NoneType'>
````

### 💻 Cell 8 (Code): Operators


````python
# 1. Arithmetic Operators
a = 5
b = 2

print("--- Arithmetic Operators ---")
print("a + b =", a + b)     # Addition (7)
print("a - b =", a - b)     # Subtraction (3)
print("a * b =", a * b)     # Multiplication (10)
print("a / b =", a / b)     # Division (Always returns float: 2.5)
print("a % b =", a % b)     # Modulo / Remainder (1)
print("a ** b =", a ** b)   # Exponentiation / Power (5^2 = 25)

# 2. Relational / Comparison Operators (Return Boolean)
print("\n--- Relational Operators ---")
print("a == b:", a == b)   # False
print("a != b:", a != b)   # True
print("a > b:", a > b)     # True
print("a <= b:", a <= b)   # False

# 3. Assignment Operators
print("\n--- Assignment Operators ---")
num = 10
num += 5    # Equivalent to num = num + 5
print("num += 5 ->", num) # 15
num *= 2    # Equivalent to num = num * 2
print("num *= 2 ->", num) # 30

# 4. Logical Operators (not, and, or)
print("\n--- Logical Operators ---")
val1 = True
val2 = False

print("NOT operator:", not val1)          # False
print("AND operator:", val1 and val2)     # False (True only if BOTH are True)
print("OR operator:", val1 or val2)       # True (True if AT LEAST ONE is True)
````

### 📝 Cell 9 (Markdown): Type Conversion vs Type Casting


## 4. Type Conversion vs Type Casting
* **Type Conversion (Implicit):** Performed automatically by Python when operating on compatible types (e.g., adding an `int` and a `float` results in a `float`).
* **Type Casting (Explicit):** Manually converting a data type using functions like `int()`, `float()`, `str()`.

```

---
### 💻 Cell 10 (Code): Type Conversion & Type Casting Examples

````python
# Implicit Conversion (Automatic)
x = 2       # int
y = 4.25    # float
sum_val = x + y  # Python automatically converts x to 2.0 (float)
print("Sum:", sum_val, "| Type:", type(sum_val)) # 6.25 | float

# Explicit Type Casting (Manual)
a = "2"     # string
b = 4.25    # float

a_converted = int(a)  # Converting string "2" to integer 2
total = a_converted + b
print("Total after casting:", total) # 6.25

# Converting number to string
num = 3.14
str_num = str(num)
print("String representation:", str_num, "| Type:", type(str_num))
````
