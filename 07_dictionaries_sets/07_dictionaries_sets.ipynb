## 1. Dictionaries in Python
Dictionaries are used to store data values in **key:value** pairs.

### Key Properties:
* Created using curly braces `{}`.
* **Unordered:** There is no index (0, 1, 2...). Elements are accessed via keys.
* **Mutable:** You can add, remove, or modify items after creation.
* **No Duplicate Keys:** Keys must be unique. If a key is repeated, the new value overwrites the old one.
* **Key Restrictions:** Keys must be **immutable** types (like Strings, Integers, Floats, Tuples). Lists and Dictionaries *cannot* be used as keys. Values can be of any data type.

```python
# Creating a Dictionary
info = {
    "name": "Apna College",
    "subjects": ["Python", "C", "Java"], # List as value
    "topics": ("dict", "set"),           # Tuple as value
    "age": 35,
    "is_adult": True,
    "marks": 94.4
}

print("Info Dictionary:", info)
print("Type:", type(info))

# Accessing Values
print("Name:", info["name"])
print("Subjects:", info["subjects"])

# Modifying and Adding values (Dictionaries are Mutable)
info["name"] = "Raghav"       # Modifying existing key
info["surname"] = "Semwal"    # Adding new key-value pair
print("Updated Info:", info)

# Creating an Empty Dictionary
empty_dict = {}
empty_dict["name"] = "Apna College"
print("Empty Dictionary after adding item:", empty_dict)
```

## 2. Nested Dictionaries & Dictionary Methods
You can place a dictionary inside another dictionary. This is called **Nesting**.

### Built-in Dictionary Methods:
1. `dict.keys()`: Returns all keys in the dictionary.
2. `dict.values()`: Returns all values in the dictionary.
3. `dict.items()`: Returns all (key, value) pairs as tuples.
4. `dict.get("key")`: Returns the value for the key. (Doesn't throw an error if the key is missing).
5. `dict.update(new_dict)`: Inserts the specified items into the dictionary.

```python
# Nested Dictionary
student = {
    "name": "Rahul Kumar",
    "scores": {
        "physics": 97,
        "chemistry": 98,
        "math": 95
    }
}

# Accessing nested values
print("Nested Value (Chemistry marks):", student["scores"]["chemistry"])

# Dictionary Methods
print("\n--- Dictionary Methods ---")

# 1. keys()
print("Keys:", student.keys())
print("Keys as list:", list(student.keys())) # Typecasting to list

# 2. values()
print("Values:", list(student.values()))

# 3. items()
print("Items (Pairs):", list(student.items()))

# 4. get() vs square brackets []
print("Get Name:", student.get("name"))
print("Get missing key:", student.get("surname")) # Returns None
# print(student["surname"]) # Uncommenting this throws a KeyError!

# 5. update()
new_info = {"city": "Delhi", "age": 20}
student.update(new_info)
print("Updated Student Dict:", student)
```

## 3. Sets in Python
A **Set** is a collection of **unordered** and **unique** elements.

### Key Properties:
* Created using curly braces `{}`.
* **Unique:** Duplicate values are automatically ignored.
* **Unordered:** Items have no defined order and no index.
* **Set is Mutable but its Elements are Immutable:** You can add/remove items from a set, but the items *inside* the set must be immutable (like ints, floats, strings, tuples). You *cannot* store a List or Dictionary inside a set (causes `TypeError: unhashable type`).

> **Important:** To create an empty set, you MUST use `set()`. Using `{}` creates an empty dictionary.

```python
# Creating a Set
collection = {1, 2, 3, 4, 4, 4, "hello", "world", "world"}

print("Set (Notice duplicates are removed & order is random):", collection)
print("Type:", type(collection))
print("Length of Set:", len(collection))

# Empty Set vs Empty Dictionary
empty_dict = {}
print("type({}):", type(empty_dict))  # <class 'dict'>

empty_set = set()
print("type(set()):", type(empty_set))  # <class 'set'>

# Sets cannot contain mutable types (Lists, Dicts)
# invalid_set = {1, 2, [3, 4]}  # Raises TypeError: unhashable type: 'list'
```

## 4. Set Methods & Operations
Sets support mathematical operations like Union and Intersection.

| Method | Description |
| :--- | :--- |
| `set.add(el)` | Adds an element |
| `set.remove(el)` | Removes element (Throws error if element doesn't exist) |
| `set.clear()` | Empties the entire set |
| `set.pop()` | Removes and returns a random element |
| `set1.union(set2)` | Combines both sets (Unique values only) |
| `set1.intersection(set2)`| Returns common elements between both sets |

```python
# Basic Set Methods
my_set = {1, 2, 3}

my_set.add(4)
my_set.add("Apna College")
print("After add():", my_set)

my_set.remove(3)
print("After remove(3):", my_set)

popped_val = my_set.pop()
print("After pop():", my_set, "| Popped value:", popped_val)

my_set.clear()
print("After clear():", my_set)

# Union and Intersection
set1 = {1, 2, 3}
set2 = {2, 3, 4}

print("\nUnion:", set1.union(set2))             # {1, 2, 3, 4}
print("Intersection:", set1.intersection(set2)) # {2, 3}
```

## 5. Practice Questions & Solutions

```python
# Q1: Store the following word meanings in a Python Dictionary:
# "table" : "a piece of furniture", "list of facts & figures"
# "cat" : "a small animal"

word_dict = {
    "cat": "a small animal",
    "table": ["a piece of furniture", "list of facts & figures"] # Using list for multiple meanings
}

print("Word Dictionary:", word_dict)
```

```python
# Q2: You are given a list of subjects for students. Assume one classroom is required for 1 subject.
# How many classrooms are needed by all students?
# List: ["python", "java", "C++", "python", "javascript", "java", "python", "java", "C++", "C"]

subjects = ["python", "java", "C++", "python", "javascript", "java", "python", "java", "C++", "C"]

# Lecture & Pythonic Method:
# Convert list to a Set to get unique values, then find its length.
unique_subjects = set(subjects)
classrooms_needed = len(unique_subjects)

print("Unique Subjects:", unique_subjects)
print("Total classrooms needed:", classrooms_needed)
```

```python
# Q3: Write a program to enter marks of 3 subjects from the user and store them in a dictionary.
# Start with an empty dictionary and add one by one.

# --- Lecture Method (using update()) ---
marks = {}

x = int(input("Enter Physics marks: "))
marks.update({"physics": x})

y = int(input("Enter Math marks: "))
marks.update({"math": y})

z = int(input("Enter Chemistry marks: "))
marks.update({"chemistry": z})

print("Marks Dictionary (Lecture Method):", marks)


# --- Alternate Pythonic Method (Direct Assignment) ---
# Direct assignment is faster, cleaner, and preferred in Python
alt_marks = {}

alt_marks["physics"] = int(input("Enter Physics marks again: "))
alt_marks["math"] = int(input("Enter Math marks again: "))
alt_marks["chemistry"] = int(input("Enter Chemistry marks again: "))

print("Marks Dictionary (Alternate Method):", alt_marks)
```

```python
# Q4: Figure out a way to store 9 and 9.0 as separate values in a set.
# (Normally, Python treats 9 and 9.0 as the exact same value in a set).

# Scenario:
normal_set = {9, 9.0}
print("Normal Set (treats 9 and 9.0 as same):", normal_set)

# Solution 1: Store one as a String
sol1_set = {9, "9.0"}
print("Solution 1 (Using String):", sol1_set)

# Solution 2 (Alternate/Better approach): Store as Tuples with their Data Type
sol2_set = {
    ("float", 9.0),
    ("int", 9)
}
print("Solution 2 (Using Tuples):", sol2_set)
```
