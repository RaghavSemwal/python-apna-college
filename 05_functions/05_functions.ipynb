## 1. Functions in Python
A **Function** is a reusable block of statements designed to perform a specific task.

### Why use Functions?
* **Reduces Redundancy:** Write the logic once and call it multiple times across the program.
* **Modularity:** Makes code clean, easy to read, test, and debug.

### Terminology:
* **Function Definition:** Defining the logic with `def function_name(param1, param2):`.
* **Parameters:** Variables passed in the function definition header that act as placeholders for input.
* **Function Call:** Executing the function using `function_name(arg1, arg2)`.
* **Arguments:** The actual values supplied during a function call.
* **Return Statement:** Sends a computed output value back to the caller (if omitted, the function returns `None`).

```python
# 1. Defining a function with parameters and return statement
def calc_sum(a, b):
    return a + b

# Function calls with arguments
sum1 = calc_sum(5, 10)
sum2 = calc_sum(12, 17)
print("Sum 1:", sum1)
print("Sum 2:", sum2)


# 2. Function with no parameters and no return value
def print_hello():
    print("Hello from Apna College!")

# Calling the function
print_hello()
print_hello()

# A function without an explicit return statement returns None
result = print_hello()
print("Return value of print_hello():", result)
```

## 2. Average of 3 Numbers Example

```python
def calc_avg(a, b, c):
    total = a + b + c
    avg = total / 3
    return avg

print("Average of 98, 97, 95:", calc_avg(98, 97, 95))
print("Average of 1, 2, 3:", calc_avg(1, 2, 3))
```

## 3. Built-in vs User-Defined Functions & Default Parameters

* **Built-in Functions:** Provided natively by Python (e.g., `print()`, `len()`, `type()`, `range()`).
  * `print()` has optional parameters like `sep=" "` (separator between items) and `end="\n"` (ending character, default is newline).
* **User-Defined Functions:** Functions created by the programmer using `def`.
* **Default Parameters:** Assigns default values to parameters when the caller does not provide an argument.
  * **Rule:** Non-default arguments **cannot** follow default arguments (default parameters must always be at the end).

```python
# 1. print() function's sep and end parameters
print("Apna", "College", sep=" - ") # Custom separator
print("Hello", end=" ")              # Ends with a space instead of a newline
print("World!")


# 2. Default Parameters
def calc_prod(a=1, b=1):
    return a * b

print("calc_prod():", calc_prod())       # Uses both defaults -> 1 * 1 = 1
print("calc_prod(4):", calc_prod(4))     # a=4, b uses default (1) -> 4 * 1 = 4
print("calc_prod(4, 3):", calc_prod(4, 3)) # a=4, b=3 -> 12

# Note: def func(a=1, b): is INVALID syntax! Default parameters must come last.
```

## 4. Functions Practice Problems

```python
# Q1: Write a function to print the length of a list (list is the parameter).

def print_len(lst):
    print("Length of list:", len(lst))

cities = ["Delhi", "Gurgaon", "Noida", "Pune", "Mumbai", "Chennai"]
heroes = ["Thor", "Ironman", "Captain America", "Shaktiman"]

print_len(cities)
print_len(heroes)
```

```python
# Q2: Write a function to print the elements of a list in a single line.

def print_list(lst):
    for item in lst:
        print(item, end=" ")
    print() # New line after finishing

heroes = ["Thor", "Ironman", "Captain America", "Shaktiman"]
print_list(heroes)

# Alternate Pythonic Method (Using * unpacking):
def print_list_alt(lst):
    print(*lst)

print_list_alt(heroes)
```

```python
# Q3: Write a function to find the factorial of n (n is the parameter).

# Iterative function (Lecture Method)
def calc_fact(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

print("Factorial of 5:", calc_fact(5))
print("Factorial of 6:", calc_fact(6))
```

```python
# Q4: Write a function to convert USD to INR (assume 1 USD = 83 INR).

def usd_to_inr(usd_val, conversion_rate=83):
    inr_val = usd_val * conversion_rate
    print(f"${usd_val} USD = ₹{inr_val} INR")

usd_to_inr(1)
usd_to_inr(100)
usd_to_inr(73)
```

```python
# Q5 (Homework): Write a function that takes a number and returns "EVEN" or "ODD".

def check_odd_even(num):
    if num % 2 == 0:
        return "EVEN"
    else:
        return "ODD"

print("14 is:", check_odd_even(14))
print("7 is:", check_odd_even(7))

# Alternate One-Liner (Ternary Operator):
def check_odd_even_alt(num):
    return "EVEN" if num % 2 == 0 else "ODD"

print("22 is (Alt):", check_odd_even_alt(22))
```

## 5. Recursion in Python
**Recursion** is a programming technique where a **function calls itself repeatedly** to solve smaller sub-problems of the original problem.

### 2 Essential Parts of Recursion:
1. **Base Case:** The stopping condition that terminates recursion and prevents infinite function calls (`RecursionError: maximum recursion depth exceeded`).
2. **Recursive Case / Relation:** The step where the function calls itself with smaller/modified input values.

### Call Stack:
Every recursive call gets pushed onto the system's **Call Stack**. Once the base case is hit, the calls resolve in reverse order (LIFO: Last-In, First-Out) and are popped off the stack.

```python
# Basic Recursive Function: Printing numbers from n down to 1
def show(n):
    # Base Case
    if n == 0:
        return
    # Work
    print(n, end=" ")
    # Recursive Call (Smaller problem: n - 1)
    show(n - 1)

print("Recursive show(5):")
show(5)
print()
```

## 6. Factorial Using Recursion
Mathematically:
$$\text{fact}(n) = n \times \text{fact}(n - 1)$$
**Base Case:** $\text{fact}(0) = 1$ and $\text{fact}(1) = 1$.

```python
# Recursive Factorial Function
def fact_rec(n):
    # Base Case
    if n == 0 or n == 1:
        return 1
    # Recursive Step
    return n * fact_rec(n - 1)

print("Recursive 5! =", fact_rec(5))
print("Recursive 4! =", fact_rec(4))
```

## 7. Recursion Practice Problems

```python
# Q1: Write a recursive function to calculate the sum of first n natural numbers.
# Mathematical Relation: sum(n) = n + sum(n - 1)
# Base Case: sum(0) = 0

def calc_sum_rec(n):
    if n == 0:
        return 0
    return n + calc_sum_rec(n - 1)

print("Sum of first 5 natural numbers (Recursive):", calc_sum_rec(5))
print("Sum of first 10 natural numbers (Recursive):", calc_sum_rec(10))
```

```python
# Q2: Write a recursive function to print all elements in a list.
# Base Case: index reaches the length of the list.

def print_list_rec(lst, idx=0):
    # Base Case
    if idx == len(lst):
        return
    # Work
    print(lst[idx], end=" ")
    # Recursive Step
    print_list_rec(lst, idx + 1)

fruits = ["Mango", "Litchi", "Apple", "Banana"]
print("Printing list recursively:")
print_list_rec(fruits)
print()
```
