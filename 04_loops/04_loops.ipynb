## 1. What are Loops?
Loops are used in programming to **repeat a block of code multiple times** until a specific condition is met.

* **Iteration:** One single execution cycle of the loop.
* **Iterator:** The variable that keeps track of the current iteration/index (commonly `i`, `j`, `count`).
* **Two main types of loops in Python:**
  1. `while` loop (Condition-based iteration)
  2. `for` loop (Sequential / Collection-based traversal)


## 2. `while` Loops
A `while` loop executes as long as a specified boolean condition evaluates to `True`.

### 3 Key Components of a `while` Loop:
1. **Initialization:** Starting value of the iterator variable (e.g., `i = 1`).
2. **Stopping Condition:** Condition checked before each iteration (e.g., `i <= 5`).
3. **Updation / Increment:** Modifying the iterator inside the loop (e.g., `i += 1`) to prevent infinite loops.

````python
# 1. Print a message multiple times
count = 1
while count <= 5:
    print("Hello World")
    count += 1

print("Loop finished, final count:", count)

# 2. Print numbers 1 to 5
i = 1
while i <= 5:
    print(i)
    i += 1

# 3. Print numbers in reverse (5 to 1)
j = 5
while j >= 1:
    print(j)
    j -= 1
````

## 3. Loop Control Keywords: `break` & `continue`
* **`break`:** Immediately **terminates** the loop and jumps to the code outside.
* **`continue`:** Skips the **rest of the current iteration** and jumps directly to the next iteration cycle.

````python
# --- Demonstration of break ---
print("--- break demonstration ---")
i = 1
while i <= 5:
    if i == 3:
        print("Breaking at i =", i)
        break
    print(i)
    i += 1
print("Loop Terminated via break.\n")


# --- Demonstration of continue ---
print("--- continue demonstration (skipping 3) ---")
j = 0
while j < 5:
    j += 1
    if j == 3:
        continue  # Skips print(3)
    print(j)


# --- Print Odd numbers using continue ---
print("\nOdd numbers between 1 and 10:")
num = 0
while num < 10:
    num += 1
    if num % 2 == 0:
        continue # Skip even numbers
    print(num)
````

## 4. `for` Loops & Sequential Traversal
`for` loops in Python are used for **sequential traversal** over collections like Lists, Tuples, Strings, and Dictionaries.

### Syntax:
```python
for element in collection:
    # Action performed on element

````python

# 1. Traversing a List
fruits = ["Apple", "Banana", "Mango", "Litchi"]
for fruit in fruits:
    print("Fruit:", fruit)

# 2. Traversing a Tuple
nums = (10, 20, 30, 40)
for n in nums:
    print("Tuple item:", n)

# 3. Traversing a String
word = "ApnaCollege"
for char in word:
    print(char, end=" ")
print()
````

## 5. `for` Loop with optional `else` Block
* The `else` block executes **only when the loop completes all iterations naturally**.
* If the loop is terminated early using **`break`**, the `else` block is **skipped**.

````python
# Normal loop completion -> else executes
for item in [1, 2, 3]:
    print("Item:", item)
else:
    print("Loop finished completely!\n")

# Early break -> else does NOT execute
for item in [1, 2, 3, 4, 5]:
    if item == 3:
        print("Found 3, breaking!")
        break
else:
    print("This will not be printed because of break.")
````

## 6. The `range()` Function
`range()` generates an immutable sequence of numbers.

### Syntax: `range(start, stop, step)`
* **`start` (optional):** Starting value (Default is `0`).
* **`stop` (required):** Generates up to `stop - 1` (stop is excluded).
* **`step` (optional):** Increments/decrements by this step size (Default is `1`).

````python



# 1. Single argument: range(stop) -> 0 up to 4
print("range(5):", list(range(5)))

# 2. Two arguments: range(start, stop) -> 2 up to 7
print("range(2, 8):", list(range(2, 8)))

# 3. Three arguments: range(start, stop, step) -> Even numbers up to 10
print("Even numbers:", list(range(2, 11, 2)))

# 4. Negative step: range(start, stop, step) -> Counts backwards
print("Reverse range:", list(range(10, 0, -1)))
````

## 7. The `pass` Statement
`pass` is a null statement. It acts as a placeholder when syntactic structure requires a block of code, but you do not want to execute any commands.

````python
# Using pass inside an empty loop or condition
for i in range(5):
    pass  # To be implemented later

if True:
    pass

print("Executed without syntax errors thanks to pass.")
````

### Problem 1: Print numbers from 1 to 100 & 100 to 1

````python
# Problem 1: Print 1 to 100 and 100 to 1

# --- Lecture Method: while loop ---
print("1 to 100 (while loop):")
i = 1
while i <= 100:
    print(i, end=" ")
    i += 1
print()

# --- Alternate / Pythonic Method: for loop with range ---
print("\n100 to 1 (for loop range):")
for num in range(100, 0, -1):
    print(num, end=" ")
print()
````

### Problem 2: Print the Multiplication Table of a number `n`

````python
# Problem 2: Multiplication Table of n
n = int(input("Enter number for table: "))

# --- Lecture Method (while loop) ---
print(f"--- Table of {n} (while loop) ---")
i = 1
while i <= 10:
    print(f"{n} x {i} = {n * i}")
    i += 1

# --- Alternate Method (for loop range) ---
print(f"\n--- Table of {n} (for loop) ---")
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
````

### Problem 3: Linear Search in a Tuple
Find the index of number `x` in a given Tuple.

````python
# Problem 3: Linear Search in a Tuple
nums = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)
x = 36

# --- Lecture Method (Manual traversal with index tracking) ---
idx = 0
found = False
while idx < len(nums):
    if nums[idx] == x:
        print(f"Found {x} at index {idx} (while loop method)")
        found = True
        break
    idx += 1

if not found:
    print(f"{x} not found in tuple.")

# --- Alternate Pythonic Method 1 (enumerate in for loop) ---
for index, val in enumerate(nums):
    if val == x:
        print(f"Found {x} at index {index} (enumerate method)")
        break

# --- Alternate Pythonic Method 2 (Built-in index method) ---
if x in nums:
    print(f"Found {x} at index {nums.index(x)} (built-in method)")
````

### Problem 4: Sum of first `n` Natural Numbers

````python
del sum
# Problem 4: Sum of first n Natural Numbers
n = 5

# --- Lecture Method (while loop) ---
total_sum = 0
i = 1
while i <= n:
    total_sum += i
    i += 1
print(f"Sum of first {n} numbers (while loop): {total_sum}")

# --- Alternate Method 1 (for loop with range) ---
total_sum_for = 0
for i in range(1, n + 1):
    total_sum_for += i
print(f"Sum using for loop: {total_sum_for}")

# --- Alternate Method 2 (Built-in sum() & range) ---
print(f"Sum using sum(range()): {sum(range(1, n + 1))}")

# --- Alternate Method 3 (Mathematical Formula: n*(n+1)//2) ---
print(f"Sum using formula n*(n+1)//2: {(n * (n + 1)) // 2}")
````

### Problem 5: Factorial of first `n` Numbers

````python
# Problem 5: Factorial of n (e.g. 5! = 1 * 2 * 3 * 4 * 5 = 120)
import math

n = 5

# --- Lecture Method (for loop) ---
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"Factorial of {n} (for loop): {fact}")

# --- Alternate Method 1 (while loop) ---
fact_while = 1
i = 1
while i <= n:
    fact_while *= i
    i += 1
print(f"Factorial of {n} (while loop): {fact_while}")

# --- Alternate Method 2 (math.factorial) ---
print(f"Factorial of {n} (math.factorial): {math.factorial(n)}")
````
