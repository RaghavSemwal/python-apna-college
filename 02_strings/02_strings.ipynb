## 1. What is a String?
A **String** is a data type in Python that stores a sequence of characters (letters, numbers, spaces, symbols).

### Ways to Define Strings:
1. Double Quotes: `"Hello"` (Most commonly used)
2. Single Quotes: `'Hello'` (Useful if string contains double quotes)
3. Triple Quotes: `"""Hello"""` (Useful for multi-line text or string containing single quotes)

```python
# Creating strings using different quotes
str1 = "This is a string."
str2 = 'This is Apna College tutorial.'
str3 = """This is a multi-line string."""

print(str1)
print(str2)
print(str3)

# Why multiple quote styles? Useful when escaping inner quotes:
# Example: Using double quotes because string contains a single quote / apostrophe
sentence = "This is Apna College's Python tutorial."
print(sentence)
```

## 2. Escape Sequence Characters
Escape sequences are special formatting characters prefixed with a backslash `\`.
* `\n` — Inserts a **New Line**
* `\t` — Inserts a **Tab Space**

```python
# \n -> New line
str_newline = "This is a string.\nWe are creating it in Python."
print(str_newline)

# \t -> Tab space
str_tab = "This is a string.\tWe are creating it in Python."
print(str_tab)
```

## 3. Basic String Operations

### A. Concatenation
Joining two or more strings together using the `+` operator.

### B. Length of a String (`len()`)
The `len()` function returns the total count of characters in a string, **including empty spaces and special symbols**.

```python
# 1. Concatenation
str1 = "मम "
str2 = "महाविद्यालयः"
final_str = str1 + str2
print("Concatenated String:", final_str)

# Concatenation with space
final_str_with_space = "अपना" + " " + "College"
print("With Space:", final_str_with_space)

# 2. Length of String
print("Length of str1:", len("अपना"))
print("Length of str2:", len("College"))
print("Length of final string (includes space):", len(final_str_with_space))
```

## 4. Indexing in Python Strings
* Every character in a string is assigned a position number called an **Index**.
* Python indexing **starts at `0`** and goes up to `length - 1`.
* **Accessing Characters:** `str[index]` returns the character at that specific index.
* **Strings are Immutable:** You cannot change individual characters in a string via assignment (e.g., `str[0] = 'A'` gives an error).

```python
# Indexing Example
str_val = "Apna College"

# Index positions:
# 'A' -> 0, 'p' -> 1, 'n' -> 2, 'a' -> 3, ' ' -> 4, 'C' -> 5 ...

print("Character at index 0:", str_val[0])
print("Character at index 1:", str_val[1])
print("Character at index 5:", str_val[5])

# Strings are Immutable!
# str_val[0] = 'B'  <-- This line will raise a TypeError: 'str' object does not support item assignment
```

## 5. String Slicing (`[start_idx : end_idx]`)
Slicing allows you to extract a sub-part (slice) of a string.

**Syntax:** `str[start_index : end_index]`
* **`start_index`** is **included**.
* **`end_index`** is **excluded** (goes up to `end_index - 1`).

### Shortcuts:
* `str[ : end]` -> Starts automatically from index `0`.
* `str[start : ]` -> Goes automatically to the very end (`len(str)`).

### Negative Indexing (Backward Counting):
* Counts from right to left starting with `-1` for the last character.
* Last char: `-1`, Second last char: `-2`, etc.

```python
str_val = "Apna College"

# Positive Slicing
print("str_val[0:4]:", str_val[0:4])   # "Apna" (indices 0, 1, 2, 3)
print("str_val[1:4]:", str_val[1:4])   # "pna"
print("str_val[5:12]:", str_val[5:12]) # "College"

# Slicing Shortcuts
print("str_val[:4]:", str_val[:4])     # Same as [0:4] -> "Apna"
print("str_val[5:]:", str_val[5:])     # Same as [5:len(str_val)] -> "College"

# Negative Indexing Slicing
# 'A'(-12) 'p'(-11) 'n'(-10) 'a'(-9) ' '(-8) 'C'(-7) 'o'(-6) 'l'(-5) 'l'(-4) 'e'(-3) 'g'(-2) 'e'(-1)
fruit = "Apple"
# A(-5) p(-4) p(-3) l(-2) e(-1)
print("\nNegative Slicing on 'Apple':")
print("fruit[-3:-1]:", fruit[-3:-1])   # "pl" (indices -3, -2)
print("fruit[-5:-2]:", fruit[-5:-2])   # "App"
```

## 6. Built-in String Functions
Python provides several useful in-built methods for strings. Note that string methods do not modify the original string (since strings are immutable); they return a **new string**.

1. **`str.endswith("substr")`** — Returns `True` if string ends with given substring, else `False`.
2. **`str.capitalize()`** — Capitalizes the 1st character of the string.
3. **`str.replace("old", "new")`** — Replaces all occurrences of `old` with `new`.
4. **`str.find("substr")`** — Returns the **first starting index** of given substring (-1 if not found).
5. **`str.count("substr")`** — Returns total number of occurrences of substring.

```python
str_text = "i am studying Python from Apna College"

# 1. endswith()
print("Ends with 'College':", str_text.endswith("College"))  # True
print("Ends with 'app':", str_text.endswith("app"))          # False

# 2. capitalize()
print("Capitalized:", str_text.capitalize())  # Capitalizes 1st character 'i' to 'I'

# 3. replace()
print("Replaced 'o' with 'a':", str_text.replace("o", "a"))
print("Replaced 'Python' with 'JavaScript':", str_text.replace("Python", "JavaScript"))

# 4. find()
print("First index of 'Python':", str_text.find("Python"))  # Returns index 14
print("First index of 'o':", str_text.find("o"))            # Returns first occurrence index
print("Search non-existing 'q':", str_text.find("q"))       # Returns -1

# 5. count()
print("Count of 'o':", str_text.count("o"))
print("Count of 'from':", str_text.count("from"))
```
