## 1. What are Conditional Statements?
Conditional statements allow us to execute specific blocks of code based on whether a condition evaluates to `True` or `False`.

### Keywords Used:
* **`if`**: Evaluates the first condition.
* **`elif`** (*else if*): Evaluates additional conditions if previous `if` or `elif` statements were `False`.
* **`else`**: Executes if **all** preceding conditions evaluate to `False`.

### Syntax & Indentation Rules:
* Conditions end with a colon (`:`).
* Python uses **Indentation** (4 spaces or 1 tab) to group statements inside a conditional block instead of curly braces `{}`.

```python
# 1. Basic if-else Example: Age Verification for Voting / Driving
age = 21

if age >= 18:
    print("Can vote")
    print("Can apply for license")
else:
    print("Cannot vote")
```

```python
# 2. if-elif-else Example: Traffic Light Simulation
light = "green"

if light == "red":
    print("Stop")
elif light == "green":
    print("Go")
elif light == "yellow":
    print("Look / Wait")
else:
    print("Light is broken")

print("End of code")
```

## 2. Student Grading System Example
A program that assigns grades based on marks:
* **Marks >= 90**: Grade A
* **90 > Marks >= 80**: Grade B
* **80 > Marks >= 70**: Grade C
* **Marks < 70**: Grade D

```python
# Student Grading System using Logical Operators
marks = int(input("Enter student marks: "))

if marks >= 90:
    grade = "A"
elif marks >= 80 and marks < 90:
    grade = "B"
elif marks >= 70 and marks < 80:
    grade = "C"
else:
    grade = "D"

print("Grade of the student:", grade)
```

## 3. Nesting in Conditional Statements
**Nesting** means writing an `if-else` statement inside another `if-else` block.

```python
# Nesting Example
age = 34

if age >= 18:
    if age >= 80:
        print("Cannot drive (upper age limit)")
    else:
        print("Can drive")
else:
    print("Cannot drive (underage)")
```

## 4. Practice Questions — Conditional Statements

```python
# Q1: Write a program to check if a number entered by the user is Odd or Even.

num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is an Even number")
else:
    print(num, "is an Odd number")
```

```python
# Q2: Write a program to find the greatest of 3 numbers entered by the user.

a = int(input("Enter first number (a): "))
b = int(input("Enter second number (b): "))
c = int(input("Enter third number (c): "))

if a >= b and a >= c:
    print("First number (a) is the largest:", a)
elif b >= c:
    print("Second number (b) is the largest:", b)
else:
    print("Third number (c) is the largest:", c)
```

```python
# Q3: Write a program to check if a number is a multiple of 7 or not.

num = int(input("Enter a number: "))

if num % 7 == 0:
    print(num, "is a multiple of 7")
else:
    print(num, "is NOT a multiple of 7")
```

```python
# Q4 (Homework Problem): Write a program to find the greatest of 4 numbers entered by the user.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))
d = int(input("Enter fourth number: "))

if a >= b and a >= c and a >= d:
    print("Largest number is:", a)
elif b >= c and b >= d:
    print("Largest number is:", b)
elif c >= d:
    print("Largest number is:", c)
else:
    print("Largest number is:", d)
```
