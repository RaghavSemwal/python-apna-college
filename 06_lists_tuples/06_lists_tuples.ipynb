## 1. Lists in Python
A **List** is a built-in sequence data type in Python that stores a collection of values.

### Key Properties:
* Declared using square brackets `[]`.
* **Heterogeneous:** Can store elements of **different data types** together (`int`, `float`, `str`, `bool`).
* **Mutable:** Elements inside a list can be modified / reassigned after creation.
* Zero-based indexing (`0` to `len - 1`).

```python
# Creating Lists
marks = [94.4, 87.5, 95.0, 66.4, 45.1]
student = ["Karan", 85, 19, "Delhi"] # Stores str, int, int, str together

print("Marks List:", marks)
print("Type of marks:", type(marks))
print("Length of marks list:", len(marks))

# Accessing elements via indexing
print("First student detail (Name):", student[0])
print("Student marks:", student[1])

# Mutability Demonstration (Modifying list items)
student[0] = "Arjun" # Allowed in Lists
print("Updated student list:", student)
```

## 2. List Slicing & Negative Indexing
Similar to strings, lists support slicing to access sub-lists.

* **Syntax:** `list_name[start_index : end_index]`
* `end_index` is **excluded**.
* Negative indices start from `-1` (last element) to `-len` (first element).

```python
marks = [85, 94, 76, 63, 48]

# Positive Slicing
print("marks[1:4]:", marks[1:4])   # Indices 1, 2, 3 -> [94, 76, 63]
print("marks[:3]:", marks[:3])     # Indices 0 to 2 -> [85, 94, 76]
print("marks[2:]:", marks[2:])     # Index 2 to end -> [76, 63, 48]

# Negative Slicing
print("marks[-3:-1]:", marks[-3:-1]) # Indices -3, -2 -> [76, 63]
```

## 3. List Methods
Methods in Python directly modify the original list in-place (because lists are mutable) and return `None`.

| Method | Description |
| :--- | :--- |
| `list.append(val)` | Adds element at the end of the list |
| `list.sort()` | Sorts the list in ascending order |
| `list.sort(reverse=True)` | Sorts the list in descending order |
| `list.reverse()` | Reverses the list elements |
| `list.insert(idx, val)` | Inserts value at the specific index |
| `list.remove(val)` | Removes the **first occurrence** of value |
| `list.pop(idx)` | Removes element at index `idx` |

```python
list_val = [2, 1, 3]

# 1. append()
list_val.append(4)
print("After append(4):", list_val) # [2, 1, 3, 4]

# 2. sort() - Ascending
list_val.sort()
print("After sort():", list_val) # [1, 2, 3, 4]

# 3. sort(reverse=True) - Descending
list_val.sort(reverse=True)
print("After sort(reverse=True):", list_val) # [4, 3, 2, 1]

# 4. reverse()
fruits = ["banana", "apple", "mango"]
fruits.reverse()
print("After reverse():", fruits)

# 5. insert(idx, val)
list_val = [2, 1, 3]
list_val.insert(1, 5) # Insert 5 at index 1
print("After insert(1, 5):", list_val) # [2, 5, 1, 3]

# 6. remove(val) & pop(idx)
list_val.remove(5) # Removes first '5'
print("After remove(5):", list_val)

popped_item = list_val.pop(2) # Removes item at index 2
print("After pop(2):", list_val, "| Popped Item:", popped_item)
```

## 4. Tuples in Python
A **Tuple** is a built-in sequence data type used to create **immutable** sequences of values.

### Key Properties:
* Declared using round parentheses `()`.
* **Immutable:** Cannot add, remove, or modify elements after creation.
* Zero-based indexing.

> **Single Element Syntax:** A single element tuple requires a trailing comma, e.g., `(1,)`. Without a comma, Python treats `(1)` as an integer.

```python
tup = (2, 1, 3, 1)
print("Tuple:", tup)
print("Type of tup:", type(tup))

# Single element vs Parenthesized integer
single_tup = (1,)  # Valid tuple
not_a_tup = (1)    # Evaluates to int

print("type((1,)):", type(single_tup))
print("type((1)):", type(not_a_tup))

# Immutability Demonstration
# tup[0] = 5  <-- Raises TypeError: 'tuple' object does not support item assignment
```

## 5. Tuple Methods
Tuples have fewer built-in methods because they cannot be modified.

1. **`tup.index(element)`**: Returns index of the first occurrence of `element`.
2. **`tup.count(element)`**: Returns count of occurrences of `element`.

```python
tup = (1, 2, 3, 2, 4, 2)

print("First index of element '2':", tup.index(2)) # Returns 1
print("Count of element '2':", tup.count(2))       # Returns 3
```

```python
# Q1: Write a program to ask the user to enter names of 3 favorite movies and store them in a list.

# Lecture Video Method:
movies = []
mov1 = input("Enter first movie: ")
mov2 = input("Enter second movie: ")
mov3 = input("Enter third movie: ")

movies.append(mov1)
movies.append(mov2)
movies.append(mov3)
print("Movies List (Lecture Method):", movies)
```

```python
# Q1 Alternate Method (Direct List Assignment / Cleaner Code):

movies = [
    input("Enter first movie: "),
    input("Enter second movie: "),
    input("Enter third movie: ")
]
print("Movies List (Alternate Method):", movies)
```

```python
# Q2: Write a program to check if a list contains a palindrome of elements or not.
# (Example: [1, 2, 3, 2, 1] or ["m", "a", "a", "m"])

# Lecture Video Method (Using copy() and reverse()):
list1 = [1, 2, 3, 2, 1]

copy_list1 = list1.copy()
copy_list1.reverse()

if copy_list1 == list1:
    print("List is a Palindrome (Lecture Method)")
else:
    print("List is NOT a Palindrome")
```

```python
# Q2 Alternate Method (Using Python Slicing Step `[::-1]`):
# In Python, `list[::-1]` creates a reversed copy directly in a single line.

list1 = [1, 2, 3, 2, 1]

if list1 == list1[::-1]:
    print("List is a Palindrome (Alternate Slicing Method)")
else:
    print("List is NOT a Palindrome")
```
